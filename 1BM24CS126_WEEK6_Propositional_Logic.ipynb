```python
import itertools

def get_alpha(A, B):
    return A or B

def get_kb(A, B, C):
    return (A or C) and (B or (not C))

combinations = list(itertools.product([False, True], repeat=3))


to_str = lambda x: "True " if x else "False"

print("A     B     C     | A∨C   B∨¬C  | KB    α     | Model Status")
print("-" * 65)

entails = True
for A, B, C in combinations:
    a_or_c = A or C
    b_or_not_c = B or (not C)
    kb = get_kb(A, B, C)
    alpha = get_alpha(A, B)

    if kb and not alpha:
        entails = False

    status = "<-- KB is TRUE" if kb else ""

    print(f"{to_str(A)} {to_str(B)} {to_str(C)} | {to_str(a_or_c)} {to_str(b_or_not_c)} | {to_str(kb)} {to_str(alpha)} | {status}")

print("\n" + "="*30)
print(f"Conclusion: KB ⊨ α is {entails}")
print("="*30)
```
