```python
print("Time Complexity of DFS: O(b^d), where b is the branching factor and d is the depth of the solution.")
print("Space Complexity of DFS: O(b*d), where b is the branching factor and d is the maximum depth of the search tree.")

def get_neighbors(state):
    neighbors = []

    zero = state.index(0)
    row = zero // 3
    col = zero % 3

    moves = [
        (-1, 0),
        (1, 0),
        (0, -1),
        (0, 1)
    ]

    for dr, dc in moves:
        new_row = row + dr
        new_col = col + dc

        if 0 <= new_row < 3 and 0 <= new_col < 3:
            new_zero = new_row * 3 + new_col

            new_state = list(state)
            new_state[zero], new_state[new_zero] = \
                new_state[new_zero], new_state[zero]

            neighbors.append(tuple(new_state))

    return neighbors


def dfs(initial, goal):
    stack = [(initial, [initial])]
    visited = set()

    while stack:
        current, path = stack.pop()

        if current == goal:
            return path

        if current in visited:
            continue

        visited.add(current)

        for neighbor in get_neighbors(current):
            if neighbor not in visited:
                stack.append((neighbor, path + [neighbor]))

    return None


def format_state(state, step_no):
    s = f"Step {step_no}:\n"
    for i in range(0, 9, 3):
        s += str(state[i:i+3]) + "\n"
    s += "\n"
    return s


initial = (2, 8, 3,
           1, 6, 4,
           7, 0, 5)

goal = (1, 2, 3,
        8, 0, 4,
        7, 6, 5)

solution = dfs(initial, goal)

if solution:
    print("DFS Solution:")
    for step_no, state in enumerate(solution):
        print(format_state(state, step_no + 1))
else:
    print("No solution found")
```
