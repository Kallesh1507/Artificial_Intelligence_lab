```python


def vacuum_world(position, left, right):

    print("\nInitial State:", (position, left, right))

    cost = 0

    while left == 'D' or right == 'D':


        if position == 'A':

            if left == 'D':
                print("Action: Pick the Dust from Left")
                left = 'C'
                print("State:", (position, left, right))
                cost += 0

            elif right == 'D':
                print("Action: Move Right")
                position = 'B'
                cost += 2
                print("State:", (position, left, right))


        elif position == 'B':

            if right == 'D':

                print("Action: Pick the Dust from Right")
                right = 'C'
                cost += 1
                print("State:", (position, left, right))

            elif left == 'D':
                print("Action: Move Left")
                position = 'A'
                cost += 2
                print("State:", (position, left, right))

    print("\nGoal State Reached!")
    print("Final State:", (position, left, right))
    print("Total Cost:", cost)

position = input("Enter vacuum position (A/B): ").upper()
left = input("Enter left room status (C/D): ").upper()
right = input("Enter right room status (C/D): ").upper()

if position not in ['A', 'B'] or left not in ['C', 'D'] or right not in ['C', 'D']:
    print("Invalid input!")
else:
    vacuum_world(position, left, right)
```
